```python
# making the training text longer to have more representative token statistics
# text from https://www.reedbeta.com/blog/programmers-intro-to-unicode/
text = """Ankara'dan İstanbul'a giden öğrenciler sabah erkenden tren istasyonunda buluştular. Türkiye'nin farklı şehirlerinden gelen öğrenciler Türkçe kitaplarını, defterlerini ve bilgisayarlarını yanlarına aldılar. Yolculuk boyunca araştırmalarını, gözlemlerini ve öğrenme deneyimlerini paylaştılar.
İstanbul'da üniversitedeki öğretmenler öğrencilerle yeni teknolojileri konuştular. Öğrencilerin çalışmaları, arkadaşlıkları ve merakları onları daha iyi sorular sormaya yöneltti. Yazdıkları yazıları düzenlediler, okudukları kitapları karşılaştırdılar ve öğrendiklerini anlattılar.
Günümüzde Türkçe doğal dil işleme araştırmalarında sözcüklerin eklerle uzaması önemli bir konudur. Kitaplarımızdan, evlerimizdeki, okullarımızdaki ve şehirlerimizden gibi sözcükler birçok anlamı bir arada taşır. Türkçenin ç, ğ, ı, ö, ş, ü harfleri ve İstanbul, İzmir, Çanakkale, Eskişehir gibi yer adları UTF-8 kodlamasında birden fazla byte gerektirir.
Bir öğrenci bilgisayarında basit bir tokenizer geliştirdi. Önce metni byte'lara ayırdı, ardından en çok tekrarlanan ikilileri buldu. Bu ikilileri yeni tokenlara dönüştürerek daha kısa diziler elde etti. Sonunda merhaba 👋 ve harika 🎉 gibi emoji içeren mesajları da sorunsuz yeniden oluşturdu.
Araştırmacılar öğrencilerin düşüncelerini, öğrenciler araştırmacıların önerilerini dikkatle dinlediler. Türkiye'deki üniversiteler ve İstanbul'daki kütüphaneler öğrenmeyi destekliyor. Ankara'dan gelen öğrenciler İstanbul'da çalışmalarını sürdürürken öğretmenlerinin görüşlerini değerlendiriyor.
"""
tokens = text.encode("utf-8") # raw bytes
tokens = list(map(int, tokens)) # convert to a list of integers in range 0..255 for convenience
```

```python
def get_stats(ids):
    counts = {}
    for pair in zip(ids, ids[1:]):
        counts[pair] = counts.get(pair, 0) + 1
    return counts

def merge(ids, pair, idx):
  newids = []
  i = 0
  while i < len(ids):
    if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
      newids.append(idx)
      i += 2
    else:
      newids.append(ids[i])
      i += 1
  return newids

# ---
vocab_size = 276 # the desired final vocabulary size
num_merges = vocab_size - 256
ids = list(tokens) # copy so we don't destroy the original list

merges = {} # (int, int) -> int
for i in range(num_merges):
  stats = get_stats(ids)
  pair = max(stats, key=stats.get)
  idx = 256 + i
  print(f"merging {pair} into a new token {idx}")
  ids = merge(ids, pair, idx)
  merges[pair] = idx
```

```python
print("tokens length:", len(tokens))
print("ids length:", len(ids))
print(f"compression ratio: {len(tokens) / len(ids):.2f}X")
```

```python
vocab = {idx: bytes([idx]) for idx in range(256)}
for (p0, p1), idx in merges.items():
    vocab[idx] = vocab[p0] + vocab[p1]

def decode(ids):
  # given ids (list of integers), return Python string
  tokens = b"".join(vocab[idx] for idx in ids)
  text = tokens.decode("utf-8", errors="replace")
  return text

print(decode([128]))
```

```python
def encode(text):
  # given a string, return list of integers (the tokens)
  tokens = list(text.encode("utf-8"))
  while len(tokens) >= 2:
    stats = get_stats(tokens)
    pair = min(stats, key=lambda p: merges.get(p, float("inf")))
    if pair not in merges:
      break # nothing else can be merged
    idx = merges[pair]
    tokens = merge(tokens, pair, idx)
  return tokens

print(encode(""))
```

```python
print(decode(encode("hello world")))
```
